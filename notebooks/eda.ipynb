## Heart Disease EDA
This notebook explores the Cleveland heart disease dataset.

```python
from pathlib import Path
import pandas as pd
data_path = Path('../data/raw/heart+disease/processed.cleveland.data')
df = pd.read_csv(data_path, header=None, na_values='?')
df.columns = ['age','sex','cp','trestbps','chol','fbs','restecg','thalach','exang','oldpeak','slope','ca','thal','target']
df['target'] = (df['target'] > 0).astype(int)
df.head()
```

```python
df['target'].value_counts(normalize=True).sort_index()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

df[['age', 'trestbps', 'chol', 'thalach', 'oldpeak']].hist(
    figsize=(12, 8),
    bins=20
)

plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```
